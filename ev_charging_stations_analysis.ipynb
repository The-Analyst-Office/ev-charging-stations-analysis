# Analyze Electric Vehicle Stations in Python


## Project Overview
This project highlights an analysis of fuel stations across the United States using a dataset containing fuel types, states, ownership, access and station counts. The goal of the analysis is to investigate and highlight patterns in station availability, ownership and public access, particularly for electric charging stations.

## Dataset Overview

This project uses a dataset of alternative fuel stations across the United States, including electric charging stations. The dataset contains information about fuel types, US states, station ownership, public or private access, and the number of stations.

Each row represents a group of stations sharing the same fuel type, state, ownership category and access type, rather than an individual station. The dataset contains 819 records and five columns.

### Data Import and Initial Exploration

```python
import pandas as pd
stations = pd.read_csv('stations.csv')
# show output
stations.head()
```

#### Analysis - Initial Data Exploration

The first five rows provide an initial overview of the dataset's structure. The data includes fuel type, state, ownership category, access type and the number of stations.

All five displayed records relate to biodiesel stations across Alabama, Arkansas and Arizona. The records show different ownership and access categories, with station counts ranging from 1 to 74.

This initial exploration helps establish how the data is organised before conducting further analysis.

### Distribution of Fuel Types

```python
stations['fuel'].value_counts()
```

#### Analysis - Distribution of Fuel Types

This shows that the dataset contains seven different alternative fuel types. Electric stations appear most frequently, accounting for 260 of the 819 records, followed by compressed natural gas with 210 records.

Hydrogen has the fewest records, with just 27.

These results show that electric and compressed natural gas categories are the most frequently represented in the dataset. However, the figures represent the number of records rather than the total number of individual stations.

### Distribution of Station Ownership

```python
stations['owner'].value_counts()
```

#### Analysis - Station Ownership

Private ownership is the most frequently represented category in the dataset, accounting for 413 of the 819 records. Government ownership follows with 288 records, while utility companies account for 108 records.

Joint ownership is the least common category, appearing in just 10 records.

These findings suggest that privately owned station groups are more frequently represented than other ownership categories. However, the record counts do not necessarily indicate which ownership category operates the largest number of individual stations.

### Distribution of Station Access

```python
stations['access'].value_counts()
```

#### Analysis - Public vs Private Access

The dataset contains 415 records classified as publicly accessible and 404 records classified as privately accessible.

This shows that public and private access categories are almost equally represented, with public access accounting for slightly more records. However, these figures represent the number of records rather than the total number of individual stations. Access type also differs from ownership: a privately owned station can still be publicly accessible.

### Descriptive Statistics of Station Counts

```python
stations['number_of_stations'].describe().head()
```

#### Analysis - Descriptive Statistics

The dataset contains 819 records, with an average of approximately 34.74 stations per record. However, the median is only 5 stations, indicating that most records contain relatively small station counts.

The maximum value of 2,423 stations is substantially higher than the median, suggesting that a small number of records contain unusually large station counts.

The standard deviation of approximately 118.67 also indicates considerable variation in station counts across the dataset.

Overall, the distribution appears to be positively skewed (right-skewed), with a relatively small number of high values increasing the mean.

### Identifying the Largest Station Counts

```python
stations.sort_values(by = 'number_of_stations', ascending = False).head()
```

#### Analysis - Largest Station Counts

The five records with the highest station counts are located in California, New York, Texas, Florida and Virginia.

All five records share three characteristics: they are electric charging stations, privately owned and publicly accessible.

California has the largest station count in a single record, with 2,423 stations, followed by New York with 1,173.

These findings showcase the prominence of privately owned, publicly accessible electric charging stations among the dataset's largest records. However, these results represent the five largest records rather than the total number of stations in each state.

### Identifying the Smallest Station Counts

```python
stations.sort_values(by = 'number_of_stations').head()
```

#### Analysis - Smallest Station Counts

The five records with the smallest station counts each contain just one station.
Unlike the five largest records, which were all privately owned, publicly accessible electric charging stations, the smallest records include a mixture of fuel types, ownership categories and access types.

This suggests that small station counts occur across different categories, rather than being associated with one particular fuel type or ownership group.
These findings also highlight the considerable variation in station counts across the dataset.

### Identifying Publicly Accessible Stations

```python
public_mask = stations['access'] == 'public'
public_mask
```

#### Analysis - Identifying Publicly Accessible Stations

This Boolean mask identifies records classified as publicly accessible.
This helps to isolate publicly accessible stations for further analysis. The mask itself does not filter the dataset; it identifies which records meet the specified condition.

### Identifying Electric Charging Stations

```python
electric_stations_mask = stations['fuel'] == 'electric'
electric_stations_mask
```

#### Analysis - Identifying Electric Charging Stations

This Boolean mask identifies records associated with electric charging stations.
Of the 819 records in the dataset, 260 relate to electric charging stations. This mask will allow the analysis to focus specifically on electric charging infrastructure when combined with other filtering conditions.

### Filtering Publicly Accessible Electric Charging Stations

```python
public_electric_mask = public_mask & electric_stations_mask
public_electric = stations[public_electric_mask]
public_electric
```

#### Analysis - Publicly Accessible Electric Charging Stations

Two Boolean masks were combined to identify records that represent electric charging stations with public access.

The filtered DataFrame, public_electric, contains 134 records out of the original 819. All records in this subset have electric as their fuel type and public as their access category. This filtering step creates a more focused dataset for investigating the ownership, distribution and availability of publicly accessible electric charging stations across the United States.

### Smallest Publicly Accessible Electric Station Counts

```python
public_electric.sort_values(by = 'number_of_stations').head()
```

#### Analysis - Smallest Public Electric Station Counts

The five records with the smallest station counts in the public_electric dataset each contain one station.
Although all these records represent publicly accessible electric charging stations, their ownership categories vary.

This demonstrates that publicly accessible electric charging infrastructure is not limited to a single type of owner. It also highlights the variation in station counts within the filtered dataset.

These findings provide a useful starting point for examining how ownership relates to the distribution of publicly accessible electric charging stations.

### Identifying Privately Owned Public Electric Charging Stations

```python
private = public_electric['owner'] == 'private'
private
```

#### Analysis - Identifying Privately Owned Stations

This Boolean mask identifies privately owned records within the publicly accessible electric charging station dataset.

Of the 134 publicly accessible electric charging station records, 51 are privately owned.
This step prepares the dataset for further analysis of privately owned charging infrastructure and allows comparisons with stations operated by other ownership groups.

### Filtering Privately Owned Public Electric Charging Stations

```python
privately_owned = public_electric[private]
privately_owned
```

#### Analysis - Privately Owned Public Electric Charging Stations

The Boolean mask created in the previous step was used to filter the public_electric DataFrame, producing a new dataset called privately_owned.

This dataset contains 51 records, representing privately owned electric charging stations that are accessible to the public.

The records cover all 50 US states and Washington, DC, providing nationwide coverage of this ownership and access category.

However, the number of stations varies considerably between states. For example, California has 2,423 stations in this category, compared with 1,173 in New York.
This filtered dataset provides a foundation for examining the geographical distribution of privately owned public electric charging infrastructure.

### State Distribution of Privately Owned Public Electric Stations

```python
privately_owned['state'].describe()
```

#### Analysis - State Distribution of Privately Owned Stations

This highlights an examination of the distribution of states within the privately_owned dataset.

The results show 51 records and 51 unique state or district values, indicating that each geographical area appears exactly once.

The output identifies Alaska (AK) as the most frequent value, with a frequency of 1. However, because every state or district appears only once, this does not indicate that Alaska has the largest number of charging stations.
These results confirm that the dataset includes one record for each of the 50 US states and Washington, DC, within the privately owned, publicly accessible electric charging category.

### Filtering Non-Privately Owned Public Electric Charging Stations

```python
not_private = ~private
not_privately_owned = public_electric[not_private]
not_privately_owned
```

#### Analysis - Non-Privately Owned Public Electric Charging Stations

The existing private ownership Boolean mask was reversed to identify records that are not privately owned.

This produced a new DataFrame containing 83 records from the publicly accessible electric charging station dataset.

These records represent electric charging infrastructure operated by ownership categories other than private companies, potentially including government agencies, utility companies and joint ownership arrangements.
The resulting dataset allows further investigation of non-private charging infrastructure and provides a basis for comparison with the 51 privately owned records identified earlier.

### State Distribution of Non-Privately Owned Public Electric Stations

```python
not_privately_owned['state'].describe()
```

#### Analysis - State Distribution of Non-Privately Owned Stations

This analysis examines the geographical representation of non-privately owned, publicly accessible electric charging station records.

The dataset contains 83 records covering 49 unique state or district values.

New York (NY) is identified as the most frequently occurring value, appearing in three records. This reflects the number of records associated with New York rather than the total number of charging stations located there.

Unlike the privately owned dataset, where each state or district appears exactly once, the non-privately owned dataset contains multiple records for some states. This is consistent with the different non-private ownership categories being represented separately.

These findings provide insight into the structure and geographical coverage of non-privately owned public electric charging infrastructure.

### Descriptive Statistics of Non-Privately Owned Station Counts

```python
not_privately_owned['number_of_stations'].describe()
```

#### Analysis - Descriptive Statistics of Non-Privately Owned Stations

The not_privately_owned dataset contains 83 records, with an average of approximately 17.93 stations per record.

However, the median is only 6 stations, indicating that at least half of the records contain relatively small station counts.

The 75th percentile is 17 stations, meaning that approximately 75% of records contain 17 stations or fewer.

The maximum value of 361 stations is substantially higher than both the mean and median, suggesting that a small number of records contain unusually large station counts.

The standard deviation of approximately 44.29 also indicates considerable variation in station counts.

This overall picture shows that the distribution appears to be positively skewed (right-skewed), with a relatively small number of high values increasing the mean.

### Identifying Non-Privately Owned Records With More Than 17 Stations

```python
above_17_mask = not_privately_owned['number_of_stations'] > 17
above_17_mask
```

#### Analysis - Identifying Records Above the 75th Percentile

This Boolean mask identifies non-privately owned, publicly accessible electric charging station records containing more than 17 stations.
The threshold of 17 stations was selected because it represents the 75th percentile of the number_of_stations column, as identified in the previous analysis.

This mask allows the analysis to focus on records with comparatively high station counts within the non-privately owned dataset.


### Filtering Non-Privately Owned Records With More Than 17 Stations

```python
above_17 = not_privately_owned[above_17_mask]
above_17
```

#### Analysis - Filtering Records Above the 75th Percentile

The Boolean mask created in the previous step was applied to the not_privately_owned DataFrame to identify records containing more than 17 stations.

This produced a new DataFrame containing 21 records out of the original 83 non-privately owned public electric charging station records.

These 21 records represent approximately 25.3% of the non-privately owned dataset.

The results are consistent with the 75th percentile threshold identified earlier, as approximately one-quarter of the records contain more than 17 stations.

This filtered dataset provides a basis for examining the geographical and ownership characteristics of records with relatively high station counts.

### Sorting Non-Privately Owned Records With More Than 17 Stations

```python
above_17.sort_values('number_of_stations')
```

### Analysis - Sorting Records Above the 75th Percentile

The above_17 DataFrame was sorted by number_of_stations in ascending order to examine the distribution of records containing more than 17 stations.

The filtered dataset contains 21 records, with station counts ranging from 18 to 361.

California has the largest record in this subset, with 361 government-owned, publicly accessible electric charging stations. New York follows with 145 government-owned stations, while Connecticut has 106.

The considerable difference between the smallest and largest values demonstrates that station counts vary substantially even among records exceeding the 75th percentile.

The analysis also highlights the concentration of relatively high station counts in certain state and ownership combinations within the non-privately owned dataset.

### Identifying Public Electric Charging Stations in Washington and Oregon

```python
wa_mask = public_electric['state'] == 'WA'
or_mask = public_electric['state'] == 'OR'

wa_mask
or_mask
```

#### Analysis - Identifying Washington and Oregon Records

The two Boolean masks created point out publicly accessible electric charging station records located in Washington (WA) and Oregon (OR).
These masks allow the analysis to isolate charging station records from Washington and Oregon for a more focused geographical comparison.

### Filtering Public Electric Charging Stations in Washington and Oregon

```python
wo_mask = wa_mask | or_mask
WA_or_OR = public_electric[wo_mask]
WA_or_OR
```

#### Analysis - Public Electric Charging Stations in Washington and Oregon

The Washington and Oregon Boolean masks were combined to identify publicly accessible electric charging station records located in either state.

The resulting DataFrame, WA_or_OR, contains six records: three for Washington and three for Oregon.

Both states have records representing government, private and utility company ownership.

Based on the number_of_stations values, Washington has a total of 425 publicly accessible electric charging stations, compared with 299 in Oregon.

Private ownership accounts for the largest station count in both states, with 402 stations in Washington and 282 in Oregon.

These findings suggest that privately owned infrastructure plays a significant role in publicly accessible electric charging provision in both states, with Washington having a higher station count across all three ownership categories in this dataset.

### Sorting Washington and Oregon Stations by Ownership and Station Count

```python
WA_or_OR.sort_values(by=['owner', 'number_of_stations'])
```

#### Analysis - Comparing Station Ownership in Washington and Oregon

The `WA_or_OR` DataFrame was sorted by ownership category and then by the number of stations in ascending order.

This allows station counts to be compared directly between Washington and Oregon within each ownership category.

Private ownership accounts for the largest number of publicly accessible electric charging stations in both states. Washington has 402 privately owned stations, compared with 282 in Oregon.

Government-owned station counts are considerably lower, with 19 in Washington and 16 in Oregon. Utility company ownership accounts for the smallest counts, with 4 stations in Washington and 1 in Oregon.

Washington has more stations than Oregon in all three ownership categories represented in this dataset.

The overall comparison highlights the importance of privately owned infrastructure in providing publicly accessible electric charging stations in both states.
